# Solution

### Part 1 - Loading the data

````python
import pandas as pd
import numpy as np

DATA_PATH = '../../data'  # jump twice to the workshop folder

file = f'{DATA_PATH}/titanic.csv'
df = pd.read_csv(file)
df.info(show_counts=True)  
#nr of non-missing values displayed using info() 714 non-null
````

### Part 2 - Number of missing values

````python
df['Age'].isna().sum()
# isna checks each age and lists NaN as true and numbers as false
# sum counts True
````

### Part 3 - Compute mean age

````python
import numpy as np

# Compute mean age using the Series.mean() method
mean_age = df['Age'].mean()

#Convert Age column to numpy array
age_array = df['Age'].to_numpy()

#Compute mean using np.mean()
mean_age_np = np.mean(age_array) #returns nan

#Compute mean using np.nanmeans
mean_age_np_nan = np.nanmean(age_array) #can deal with nan

print(f'Mean age using pandas: {mean_age:.3f}')
print(f'Mean age using np.mean(): {mean_age_np:.3f}')
print(f'Mean age using np.nanmean() {mean_age_np_nan:.3f}')
````

### Part 4 - replace missing ages and convert to integers

````python
#Round average age
mean_age= np.round(mean_age)

#Boolean arrays to identify and select missing observations
is_missing = df['Age'].isna()

#Update missing observations with rounded mean
df.loc[is_missing, 'Age'] = mean_age
````

````python
#Alternative is to use fillna()
#Reload data to get the original missing values

df=pd.read_csv(file)
df['Age'] = df['Age'].fillna(value=mean_age)
````

````python
#Convert the Age column to an integer
df['Age'] = df['Age'].astype(int)  #gets rid of the decimals

df['Age']
````

````python
# Alternative approach
# Replace all missing ages with the mean age using mask

#Identify all obs wirth missing age
missing = df['Age'].isna()
#Assign the average age to all missing observations
df.loc[missing, 'Age'] = mean_age

#Print nr obs with missing age
df['Age'].isna().sum()  #zero missing age
````

### Part 5 - Generate Female Indicator

````python
#Check the sex counts
df['Sex'].value_counts()
````

````python
#Generate boolean array (True/False) whether passenger is female
is_female = df['Sex'] == 'female'

#True is converted to 1, female dummy variable converted to integer
df['Female'] = is_female.astype(int)

df[['Sex', 'Female']]

#Delete original sex column, no longer needed
del df['Sex']
````

````python
#ALternative approach: Create an indicator variable female
df['Female'] = [1 if s == 'female' else 0 for s in df['Sex']]
df[['Sex', 'Female']]
````

### Part 6 Save cleaned file

````python

#Save cleaned file
file_clean = f'{DATA_PATH}/titanic-clean.csv'
df.to_csv(file_clean, sep = ',', index=False)

#Save DataFrame as CSV
#df.to_csv('titanic-clean.csv')
#df.to_excel('titanic-clean.xlsx')
````

````python
# pct_change is change compared to last row so changes relative to the previous year

#Compute GDP growth as percentage change in GDP vs the previous year
df["GDP_growth"] = df["GDP"].pct_change()*100
````

````python
#GDP growth greater than 5%
mask = df["GDP_growth"] >= 5
df.loc[mask]
````

````python
#Alternatively select high growth episodes using query()
df.query('GDP_growth > 5')
````
